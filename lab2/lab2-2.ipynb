# Завдання 1 
Звантажити та відкрити (вручну або через запропонований скрипт на сайті) наступний датасет: Individual Household Electric Power Consumption Dataset 

```python
import pandas as pd
import numpy as np
import timeit
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python

def load_initial_data(file_path):
    df = pd.read_csv(file_path, sep=';', low_memory=False, na_values=['?'])
    return df

file_name = 'household_power_consumption.txt'
df = load_initial_data(file_name)

print(f"Кількість рядків до очищення: {len(df)}")

print("Перші 5 рядків датасету:")
df.head()
```

# Завдання 2
Здійснити data cleaning

```python
def clean_power_data(df):
    df_cleaned = df.dropna().copy()
    

    float_cols = df_cleaned.columns[2:] 
    for col in float_cols:
        df_cleaned[col] = pd.to_numeric(df_cleaned[col])
        
    df_cleaned['DateTime'] = pd.to_datetime(df_cleaned['Date'] + ' ' + df_cleaned['Time'], dayfirst=True)
    
    return df_cleaned

print(f"Кількість рядків після очищення: {len(df)}")
```

# Завдання 3
## Окремими функціями сформувати вибірки:

**Обрати всі записи, у яких загальна активна споживана потужність перевищує 5 кВт**

```python
def filter_by_active_power(df):
    return df[df['Global_active_power'] > 5.0]

execution_time = timeit.timeit(lambda: filter_by_active_power(df), number=1)

result_power = filter_by_active_power(df)

print(f"Завдання виконано за: {execution_time:.6f} секунд")
print(f"Кількість знайдених записів: {len(result_power)}")

result_power.head()
```

**Обрати всі записи, у яких сила струму лежить в межах 19-20 А, для них виявити ті, у яких пральна машина та холодильних споживають більше, ніж бойлер та кондиціонер**

```python
def filter_by_intensity_and_consumption(df):
    
    return df[(df['Global_intensity'] >= 19) & 
              (df['Global_intensity'] <= 20) & 
              (df['Sub_metering_2'] > df['Sub_metering_3'])]

execution_time = timeit.timeit(lambda: filter_by_intensity_and_consumption(df), number=1)

result_intensity = filter_by_intensity_and_consumption(df)

print(f"Завдання виконано за: {execution_time:.6f} секунд")
print(f"Кількість знайдених записів: {len(result_intensity)}")

result_intensity.head()
```

**Обрати випадковим чином 500000 записів (без повторів елементів вибірки), для них обчислити середні величини усіх 3-х груп споживання електричної енергії**

```python
def get_random_sample_stats(df, n_samples=500000):
    
    sample_df = df.sample(n=n_samples, replace=False)
    
    means = sample_df[['Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']].mean()
    
    return means

execution_time = timeit.timeit(lambda: get_random_sample_stats(df), number=1)
result_means = get_random_sample_stats(df)

print(f"Завдання виконано за: {execution_time:.6f} секунд")
print("\nСередні величини споживання для 500 000 випадкових записів:")
print(result_means)
```

**Обрати ті записи, які після 18-00 споживають понад 6 кВт за хвилину в середньому, серед відібраних визначити ті, у яких основне споживання електроенергії у вказаний проміжок часу припадає на пральну машину, сушарку, холодильник та освітлення (група 2 є найбільшою), а потім обрати кожен третій результат із першої половини та кожен четвертий результат із другої половини**

```python
def final_complex_filter(df):
    initial_filter = df[(df['Time'] > '18:00:00') & 
                        (df['Global_active_power'] > 6.0)].copy()
    
    mask_group2 = (initial_filter['Sub_metering_2'] > initial_filter['Sub_metering_1']) & \
                  (initial_filter['Sub_metering_2'] > initial_filter['Sub_metering_3'])
    
    res = initial_filter[mask_group2]
    
    n = len(res)
    if n == 0: return "Результатів не знайдено"
    
    half = n // 2
    first_part = res.iloc[:half:3]
    second_part = res.iloc[half::4]
    
    return pd.concat([first_part, second_part])

execution_time = timeit.timeit(lambda: final_complex_filter(df), number=1)
result = final_complex_filter(df)

print(f"Завдання виконано за: {execution_time:.6f} секунд")
print(f"Кількість результатів: {len(result)}")
result.head()
```

# Завдання 4
Пронормувати та стандартизувати вибраний датасет

```python
def apply_scaling(df):
    columns_to_scale = [
        'Global_active_power', 'Global_intensity', 
        'Voltage', 'Sub_metering_1', 
        'Sub_metering_2', 'Sub_metering_3'
    ]
    data = df[columns_to_scale]
    
    scaler_minmax = MinMaxScaler()
    normalized_array = scaler_minmax.fit_transform(data)
    
    df_norm = pd.DataFrame(normalized_array, columns=columns_to_scale)

    scaler_std = StandardScaler()
    standardized_array = scaler_std.fit_transform(data)
    df_std = pd.DataFrame(standardized_array, columns=columns_to_scale)

    return df_norm, df_std

norm_results, std_results = apply_scaling(df)

print("Приклад нормалізованих даних (діапазон 0-1):")
display(norm_results.head(3))

print("\nПриклад стандартизованих даних (середнє 0, відхилення 1):")
display(std_results.head(3))
```

# Завдання 5
Підрахувати коефіцієнт Пірсона та Спірмена для двох integer/real атрибутів

```python
def analyze_correlation(df, col1, col2):
    pearson = df[col1].corr(df[col2], method='pearson')
    spearman = df[col1].corr(df[col2], method='spearman')
    
    return pearson, spearman

attr1 = 'Global_active_power'
attr2 = 'Global_intensity'

p_corr, s_corr = analyze_correlation(df, attr1, attr2)

print(f"Коефіцієнт Пірсона: {p_corr:.5f}")
print(f"Коефіцієнт Спірмена: {s_corr:.5f}")
```

# Завдання 6
Провести One Hot Encoding категоріального атрибута

```python
def add_weekend_category(df):
    temp_date = pd.to_datetime(df['Date'], dayfirst=True)
    
    df_cat = df.copy()
    df_cat['Day_Type'] = np.where(temp_date.dt.weekday >= 5, 'Weekend', 'Weekday')
    
    ohe_days = pd.get_dummies(df_cat[['Date', 'Day_Type', 'Global_active_power']].head(1000), 
                              columns=['Day_Type'], 
                              prefix='Is')
    
    return ohe_days

weekend_ohe = add_weekend_category(df)

display(weekend_ohe.head(10))
```
